```python
# ================================================================
# FUTURE INTERNS - MACHINE LEARNING TASK 3 (2026)
# RESUME / CANDIDATE SCREENING SYSTEM
#
# Complete Jupyter Notebook - Single Cell Version
# ================================================================


# ================================================================
# 1. INSTALL REQUIRED LIBRARIES
# ================================================================

import sys
import subprocess
import os
import re
import glob
import warnings
from collections import Counter

warnings.filterwarnings("ignore")


def install_if_missing(import_name, package_name):

    try:
        __import__(import_name)

    except ImportError:

        print(f"Installing {package_name}...")

        subprocess.check_call(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                package_name
            ]
        )


install_if_missing("pandas", "pandas")
install_if_missing("numpy", "numpy")
install_if_missing("sklearn", "scikit-learn")
install_if_missing("matplotlib", "matplotlib")
install_if_missing("seaborn", "seaborn")
install_if_missing("bs4", "beautifulsoup4")
install_if_missing("nltk", "nltk")


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from bs4 import BeautifulSoup

import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


print("=" * 80)
print("LIBRARIES LOADED SUCCESSFULLY")
print("=" * 80)


# ================================================================
# 3. DOWNLOAD NLTK RESOURCES
# ================================================================

print("\nChecking NLP resources...")

try:
    nltk.download(
        "stopwords",
        quiet=True
    )

    nltk.download(
        "wordnet",
        quiet=True
    )

    nltk.download(
        "omw-1.4",
        quiet=True
    )

except Exception as e:

    print(
        "NLTK download warning:",
        e
    )


# ================================================================
# 4. LOAD NLTK TOOLS
# ================================================================

try:

    STOP_WORDS = set(
        stopwords.words("english")
    )

except Exception:

    STOP_WORDS = set()

LEMMATIZER = WordNetLemmatizer()


# ================================================================
# 5. FIND Resume.csv AUTOMATICALLY
# ================================================================

print("\n" + "=" * 80)
print("SEARCHING FOR Resume.csv")
print("=" * 80)


# Current Jupyter directory
current_directory = os.getcwd()

print(
    "\nCurrent Jupyter working directory:"
)

print(
    current_directory
)


# List of possible locations
possible_files = []


# ------------------------------------------------
# Location 1: Current working directory
# ------------------------------------------------

direct_file = os.path.join(
    current_directory,
    "Resume.csv"
)

if os.path.isfile(direct_file):

    possible_files.append(
        direct_file
    )


# ------------------------------------------------
# Location 2: Search subfolders
# ------------------------------------------------

try:

    recursive_files = glob.glob(
        os.path.join(
            current_directory,
            "**",
            "Resume.csv"
        ),
        recursive=True
    )

    for file in recursive_files:

        if file not in possible_files:

            possible_files.append(
                file
            )

except Exception:

    pass


# ------------------------------------------------
# Location 3: Look in common Downloads folder
# ------------------------------------------------

home_directory = os.path.expanduser(
    "~"
)

common_locations = [

    os.path.join(
        home_directory,
        "Downloads",
        "Resume.csv"
    ),

    os.path.join(
        home_directory,
        "Desktop",
        "Resume.csv"
    ),

    os.path.join(
        home_directory,
        "Documents",
        "Resume.csv"
    )

]


for file in common_locations:

    if os.path.isfile(file):

        if file not in possible_files:

            possible_files.append(
                file
            )


# ================================================================
# 6. FILE PICKER IF FILE IS NOT FOUND
# ================================================================

DATASET_PATH = None


if len(possible_files) > 0:

    DATASET_PATH = possible_files[0]

    print(
        "\n✓ Resume.csv found automatically!"
    )

    print(
        "\nFile location:"
    )

    print(
        DATASET_PATH
    )


else:

    print(
        "\nResume.csv was not found automatically."
    )

    print(
        "\nA file selection window will now open."
    )

    try:

        import tkinter as tk

        from tkinter import filedialog

        root = tk.Tk()

        root.withdraw()

        root.attributes(
            "-topmost",
            True
        )

        selected_file = (
            filedialog.askopenfilename(
                title="Select Resume.csv",
                filetypes=[
                    (
                        "CSV files",
                        "*.csv"
                    ),
                    (
                        "All files",
                        "*.*"
                    )
                ]
            )
        )

        root.destroy()

        if selected_file:

            DATASET_PATH = selected_file

            print(
                "\n✓ File selected successfully!"
            )

            print(
                "\nSelected file:"
            )

            print(
                DATASET_PATH
            )

        else:

            raise FileNotFoundError(
                "No file was selected."
            )

    except Exception as e:

        print(
            "\nFile picker could not be used."
        )

        print(
            "\nError:",
            e
        )

        print(
            "\nPlease place Resume.csv in:"
        )

        print(
            current_directory
        )

        raise FileNotFoundError(
            "\nResume.csv could not be located."
        )


# ================================================================
# 7. LOAD DATASET
# ================================================================

print("\n" + "=" * 80)
print("LOADING DATASET")
print("=" * 80)


try:

    df = pd.read_csv(
        DATASET_PATH,
        encoding="utf-8",
        low_memory=False
    )

except UnicodeDecodeError:

    print(
        "UTF-8 decoding failed."
    )

    print(
        "Trying latin-1 encoding..."
    )

    df = pd.read_csv(
        DATASET_PATH,
        encoding="latin-1",
        low_memory=False
    )

except Exception as e:

    raise Exception(
        f"\nUnable to read Resume.csv.\n\n"
        f"File: {DATASET_PATH}\n\n"
        f"Error: {e}"
    )


print(
    "\n✓ Dataset loaded successfully!"
)

print(
    "\nDataset location:"
)

print(
    DATASET_PATH
)

print(
    "\nNumber of rows:",
    df.shape[0]
)

print(
    "Number of columns:",
    df.shape[1]
)

print(
    "\nColumns:"
)

print(
    df.columns.tolist()
)


# ================================================================
# 8. VALIDATE DATASET
# ================================================================

print("\n" + "=" * 80)
print("VALIDATING DATASET")
print("=" * 80)


required_columns = [
    "ID",
    "Resume_str",
    "Category"
]


missing_columns = []

for column in required_columns:

    if column not in df.columns:

        missing_columns.append(
            column
        )


if missing_columns:

    print(
        "\nRequired columns missing:"
    )

    print(
        missing_columns
    )

    print(
        "\nAvailable columns:"
    )

    print(
        df.columns.tolist()
    )

    raise ValueError(
        "The uploaded CSV does not have the required Resume Dataset structure."
    )


print(
    "\n✓ Dataset structure is correct."
)

print(
    "✓ ID column found."
)

print(
    "✓ Resume_str column found."
)

print(
    "✓ Category column found."
)


# ================================================================
# 9. CLEAN MISSING VALUES
# ================================================================

df["Resume_str"] = (
    df["Resume_str"]
    .fillna("")
    .astype(str)
)

df["Category"] = (
    df["Category"]
    .fillna("Unknown")
    .astype(str)
)


# ================================================================
# 10. BASIC DATA EXPLORATION
# ================================================================

print("\n" + "=" * 80)
print("DATASET OVERVIEW")
print("=" * 80)


print(
    "\nFirst 5 records:"
)

display(
    df.head()
)


print(
    "\nDataset shape:"
)

print(
    df.shape
)


print(
    "\nMissing values:"
)

print(
    df.isnull().sum()
)


print(
    "\nNumber of unique categories:"
)

print(
    df["Category"].nunique()
)


print(
    "\nResume category distribution:"
)

display(
    df["Category"]
    .value_counts()
    .to_frame("Number of Resumes")
)


# ================================================================
# 11. TEXT CLEANING FUNCTION
# ================================================================

def clean_text(text):

    if text is None:

        return ""

    if pd.isna(text):

        return ""

    text = str(text)


    # Remove HTML tags
    try:

        text = BeautifulSoup(
            text,
            "html.parser"
        ).get_text(
            separator=" "
        )

    except Exception:

        pass


    # Convert HTML entities
    text = (
        text
        .replace("&amp;", " and ")
        .replace("&nbsp;", " ")
        .replace("&lt;", " ")
        .replace("&gt;", " ")
    )


    # Remove URLs
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )


    # Remove email addresses
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        " ",
        text
    )


    # Lowercase
    text = text.lower()


    # Normalize common symbols
    text = text.replace(
        "machine-learning",
        "machine learning"
    )

    text = text.replace(
        "data-science",
        "data science"
    )

    text = text.replace(
        "data-analysis",
        "data analysis"
    )

    text = text.replace(
        "power-bi",
        "power bi"
    )

    text = text.replace(
        "scikit learn",
        "scikit-learn"
    )


    # Keep alphanumeric characters and useful symbols
    text = re.sub(
        r"[^a-zA-Z0-9+#.\s]",
        " ",
        text
    )


    # Remove repeated spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    )


    return text.strip()


# ================================================================
# 12. CLEAN ALL RESUMES
# ================================================================

print("\n" + "=" * 80)
print("CLEANING RESUME TEXT")
print("=" * 80)


df["clean_resume"] = (
    df["Resume_str"]
    .apply(clean_text)
)


# Replace empty text with category text if necessary
empty_mask = (
    df["clean_resume"]
    .str.strip()
    .eq("")
)


df.loc[
    empty_mask,
    "clean_resume"
] = (
    df.loc[
        empty_mask,
        "Category"
    ]
    .astype(str)
)


print(
    "\n✓ Resume text cleaning completed."
)


# ================================================================
# 13. NLP PREPROCESSING
# ================================================================

def preprocess_text(text):

    if not text:

        return ""

    text = str(text).lower()


    # Preserve important multi-word skills before tokenization
    replacements = {

        "machine learning":
            "machine_learning",

        "deep learning":
            "deep_learning",

        "data science":
            "data_science",

        "data analysis":
            "data_analysis",

        "data analytics":
            "data_analytics",

        "natural language processing":
            "natural_language_processing",

        "computer vision":
            "computer_vision",

        "predictive modeling":
            "predictive_modeling",

        "feature engineering":
            "feature_engineering",

        "data visualization":
            "data_visualization",

        "power bi":
            "power_bi",

        "google cloud":
            "google_cloud",

        "artificial intelligence":
            "artificial_intelligence",

        "project management":
            "project_management",

        "problem solving":
            "problem_solving",

        "time management":
            "time_management",

        "human resources":
            "human_resources",

        "talent acquisition":
            "talent_acquisition",

        "network security":
            "network_security",

        "ethical hacking":
            "ethical_hacking",

        "penetration testing":
            "penetration_testing",

        "burp suite":
            "burp_suite",

        "data preprocessing":
            "data_preprocessing",

        "model evaluation":
            "model_evaluation",

        "scikit-learn":
            "scikit_learn"
    }


    for old, new in replacements.items():

        text = text.replace(
            old,
            new
        )


    # Tokenization
    tokens = re.findall(
        r"[a-zA-Z0-9_+#.-]+",
        text
    )


    processed_tokens = []


    for token in tokens:

        # Restore spaces in protected terms
        token_for_check = token.replace(
            "_",
            " "
        )


        # Stopword removal
        if (
            token_for_check in STOP_WORDS
            and "_" not in token
        ):

            continue


        # Remove very short tokens
        if (
            len(token) <= 2
            and token.lower()
            not in [
                "c",
                "r"
            ]
        ):

            continue


        # Lemmatization
        try:

            lemma = (
                LEMMATIZER
                .lemmatize(
                    token
                )
            )

        except Exception:

            lemma = token


        processed_tokens.append(
            lemma
        )


    return " ".join(
        processed_tokens
    )


print(
    "\nApplying NLP preprocessing..."
)


df["processed_resume"] = (
    df["clean_resume"]
    .apply(
        preprocess_text
    )
)


print(
    "✓ NLP preprocessing completed."
)


# ================================================================
# 14. SKILL DATABASE
# ================================================================

SKILL_LIST = [

    # Programming
    "python",
    "java",
    "c++",
    "c#",
    "javascript",
    "typescript",
    "r",
    "php",
    "ruby",
    "scala",
    "kotlin",
    "swift",
    "go",

    # Web
    "html",
    "css",
    "react",
    "angular",
    "vue",
    "node.js",
    "node",
    "django",
    "flask",
    "fastapi",
    "spring",
    "bootstrap",
    "tailwind",

    # AI / ML
    "artificial intelligence",
    "machine learning",
    "deep learning",
    "data science",
    "data analysis",
    "data analytics",
    "natural language processing",
    "nlp",
    "computer vision",
    "statistics",
    "predictive modeling",
    "feature engineering",
    "model evaluation",
    "data preprocessing",

    # ML Libraries
    "scikit-learn",
    "tensorflow",
    "keras",
    "pytorch",
    "pandas",
    "numpy",
    "matplotlib",
    "seaborn",
    "opencv",

    # Databases
    "sql",
    "mysql",
    "postgresql",
    "oracle",
    "mongodb",
    "nosql",
    "sqlite",
    "redis",

    # Cloud
    "aws",
    "azure",
    "google cloud",
    "gcp",
    "docker",
    "kubernetes",

    # Big Data
    "hadoop",
    "spark",
    "hive",
    "kafka",
    "pig",

    # BI
    "power bi",
    "tableau",
    "excel",
    "looker",
    "data visualization",

    # Cybersecurity
    "cybersecurity",
    "network security",
    "ethical hacking",
    "penetration testing",
    "owasp",
    "burp suite",
    "wireshark",
    "siem",

    # HR
    "human resources",
    "hr",
    "recruitment",
    "talent acquisition",
    "employee relations",
    "payroll",
    "training",
    "hris",

    # Soft skills
    "communication",
    "teamwork",
    "problem solving",
    "leadership",
    "critical thinking",
    "time management",
    "project management",

    # Engineering
    "autocad",
    "solidworks",
    "matlab",
    "simulation"
]


# Remove duplicates
SKILL_LIST = list(
    dict.fromkeys(
        SKILL_LIST
    )
)


print(
    "\nSkill database created."
)

print(
    "Number of skills:",
    len(SKILL_LIST)
)


# ================================================================
# 15. SKILL EXTRACTION
# ================================================================

def extract_skills(text):

    if not text:

        return []

    text = str(
        text
    ).lower()


    found_skills = []


    for skill in SKILL_LIST:

        skill_lower = skill.lower()


        # Exact phrase / word matching
        pattern = (
            r"(?<!\w)"
            +
            re.escape(
                skill_lower
            )
            +
            r"(?!\w)"
        )


        if re.search(
            pattern,
            text
        ):

            found_skills.append(
                skill
            )


    return sorted(
        set(
            found_skills
        )
    )


# ================================================================
# 16. EXTRACT RESUME SKILLS
# ================================================================

print("\n" + "=" * 80)
print("EXTRACTING SKILLS FROM RESUMES")
print("=" * 80)


df["skills"] = (
    df["clean_resume"]
    .apply(
        extract_skills
    )
)


df["skill_count"] = (
    df["skills"]
    .apply(
        len
    )
)


print(
    "✓ Skill extraction completed."
)


# ================================================================
# 17. DISPLAY SAMPLE SKILL EXTRACTION
# ================================================================

print(
    "\nSample skill extraction:"
)


for i in range(
    min(
        5,
        len(df)
    )
):

    print(
        "\n" + "-" * 70
    )

    print(
        "Candidate ID:",
        df.iloc[i]["ID"]
    )

    print(
        "Category:",
        df.iloc[i]["Category"]
    )

    print(
        "Skills:"
    )


    if df.iloc[i]["skills"]:

        print(
            ", ".join(
                df.iloc[i]["skills"]
            )
        )

    else:

        print(
            "No skills detected"
        )


# ================================================================
# 18. SKILL FREQUENCY
# ================================================================

all_skills = []


for skills in df["skills"]:

    all_skills.extend(
        skills
    )


skill_counter = Counter(
    all_skills
)


skill_frequency_df = pd.DataFrame(
    skill_counter.most_common(
        20
    ),
    columns=[
        "Skill",
        "Frequency"
    ]
)


print(
    "\n" + "=" * 80
)

print(
    "MOST COMMON SKILLS"
)

print(
    "=" * 80
)


display(
    skill_frequency_df
)


# ================================================================
# 19. COMMON SKILLS GRAPH
# ================================================================

if not skill_frequency_df.empty:

    plt.figure(
        figsize=(14, 8)
    )

    sns.barplot(
        data=skill_frequency_df,
        x="Frequency",
        y="Skill"
    )

    plt.title(
        "Top 20 Skills Found in Resumes",
        fontsize=16
    )

    plt.xlabel(
        "Number of Resumes"
    )

    plt.ylabel(
        "Skill"
    )

    plt.tight_layout()

    plt.show()


# ================================================================
# 20. CATEGORY GRAPH
# ================================================================

category_counts = (
    df["Category"]
    .value_counts()
    .head(20)
)


plt.figure(
    figsize=(14, 9)
)


sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)


plt.title(
    "Resume Category Distribution",
    fontsize=16
)


plt.xlabel(
    "Number of Resumes"
)


plt.ylabel(
    "Resume Category"
)


plt.tight_layout()


plt.show()


# ================================================================
# 21. JOB DESCRIPTION
# ================================================================

JOB_DESCRIPTION = """

We are looking for a Data Scientist to join our organization.

The candidate should have strong knowledge of Python,
Machine Learning, Data Science, Data Analysis, Statistics,
SQL and Data Visualization.

The candidate should have practical experience with
Pandas, NumPy, Scikit-learn, Matplotlib and Seaborn.

Knowledge of Natural Language Processing, Predictive Modeling,
Feature Engineering, Model Evaluation and Data Preprocessing
will be preferred.

Experience with Artificial Intelligence and Deep Learning
will be an advantage.

Strong Communication, Problem Solving and Teamwork skills
are required.

The candidate should have the ability to work with data,
build predictive models and communicate analytical results.
"""


print(
    "\n" + "=" * 80
)

print(
    "JOB DESCRIPTION"
)

print(
    "=" * 80
)

print(
    JOB_DESCRIPTION
)


# ================================================================
# 22. CLEAN JOB DESCRIPTION
# ================================================================

clean_job_description = clean_text(
    JOB_DESCRIPTION
)


processed_job_description = preprocess_text(
    clean_job_description
)


# ================================================================
# 23. EXTRACT REQUIRED JOB SKILLS
# ================================================================

job_skills = extract_skills(
    clean_job_description
)


print(
    "\n" + "=" * 80
)

print(
    "REQUIRED JOB SKILLS"
)

print(
    "=" * 80
)


for skill in job_skills:

    print(
        "✓",
        skill
    )


print(
    "\nTotal required skills:",
    len(job_skills)
)


# ================================================================
# 24. TF-IDF VECTORIZATION
# ================================================================

print(
    "\n" + "=" * 80
)

print(
    "TF-IDF VECTORIZATION"
)

print(
    "=" * 80
)


documents = (
    df["processed_resume"]
    .tolist()
    +
    [
        processed_job_description
    ]
)


vectorizer = TfidfVectorizer(

    stop_words="english",

    ngram_range=(
        1,
        2
    ),

    max_features=15000,

    sublinear_tf=True,

    min_df=1
)


tfidf_matrix = vectorizer.fit_transform(
    documents
)


resume_vectors = (
    tfidf_matrix[:-1]
)


job_vector = (
    tfidf_matrix[-1]
)


print(
    "TF-IDF matrix shape:",
    tfidf_matrix.shape
)


print(
    "✓ TF-IDF vectorization completed."
)


# ================================================================
# 25. COSINE SIMILARITY
# ================================================================

print(
    "\n" + "=" * 80
)

print(
    "CALCULATING RESUME-JOB SIMILARITY"
)

print(
    "=" * 80
)


similarity_scores = cosine_similarity(
    resume_vectors,
    job_vector
).flatten()


df["similarity_score"] = (
    similarity_scores
    *
    100
)


print(
    "✓ Cosine similarity calculated."
)


# ================================================================
# 26. SKILL MATCHING
# ================================================================

def calculate_skill_match(
    candidate_skills,
    required_skills
):

    candidate_set = set(
        candidate_skills
    )

    required_set = set(
        required_skills
    )


    if len(
        required_set
    ) == 0:

        return 0.0


    matched_skills = (
        candidate_set
        &
        required_set
    )


    return (
        len(
            matched_skills
        )
        /
        len(
            required_set
        )
        *
        100
    )


df["skill_match_score"] = (
    df["skills"]
    .apply(
        lambda skills:
        calculate_skill_match(
            skills,
            job_skills
        )
    )
)


print(
    "✓ Skill matching completed."
)


# ================================================================
# 27. MATCHED SKILLS
# ================================================================

def get_matched_skills(
    candidate_skills,
    required_skills
):

    return sorted(
        set(
            candidate_skills
        )
        &
        set(
            required_skills
        )
    )


df["matched_skills"] = (
    df["skills"]
    .apply(
        lambda skills:
        get_matched_skills(
            skills,
            job_skills
        )
    )
)


# ================================================================
# 28. MISSING SKILLS
# ================================================================

def get_missing_skills(
    candidate_skills,
    required_skills
):

    return sorted(
        set(
            required_skills
        )
        -
        set(
            candidate_skills
        )
    )


df["missing_skills"] = (
    df["skills"]
    .apply(
        lambda skills:
        get_missing_skills(
            skills,
            job_skills
        )
    )
)


print(
    "✓ Skill-gap analysis completed."
)


# ================================================================
# 29. FINAL WEIGHTED SCORE
# ================================================================

SIMILARITY_WEIGHT = 0.60

SKILL_WEIGHT = 0.40


df["final_score"] = (

    SIMILARITY_WEIGHT
    *
    df["similarity_score"]

    +

    SKILL_WEIGHT
    *
    df["skill_match_score"]
)


print(
    "\nFinal scoring formula:"
)

print(
    "Final Score = "
    "(60% × TF-IDF Similarity) + "
    "(40% × Skill Match)"
)


# ================================================================
# 30. RECOMMENDATION
# ================================================================

def get_recommendation(
    score
):

    if score >= 75:

        return "Strong Match"

    elif score >= 55:

        return "Good Match"

    elif score >= 35:

        return "Moderate Match"

    else:

        return "Low Match"


df["recommendation"] = (
    df["final_score"]
    .apply(
        get_recommendation
    )
)


# ================================================================
# 31. RANK CANDIDATES
# ================================================================

df = (
    df.sort_values(
        by="final_score",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


df["Rank"] = (
    df.index
    +
    1
)


print(
    "\n✓ All candidates ranked successfully."
)


# ================================================================
# 32. FINAL RANKING TABLE
# ================================================================

print(
    "\n" + "=" * 80
)

print(
    "TOP 20 CANDIDATES"
)

print(
    "=" * 80
)


ranking = df[
    [
        "Rank",
        "ID",
        "Category",
        "similarity_score",
        "skill_match_score",
        "final_score",
        "recommendation"
    ]
].copy()


ranking[
    "similarity_score"
] = ranking[
    "similarity_score"
].round(
    2
)


ranking[
    "skill_match_score"
] = ranking[
    "skill_match_score"
].round(
    2
)


ranking[
    "final_score"
] = ranking[
    "final_score"
].round(
    2
)


display(
    ranking.head(
        20
    )
)


# ================================================================
# 33. DETAILED TOP 10 ANALYSIS
# ================================================================

print(
    "\n" + "=" * 80
)

print(
    "DETAILED TOP 10 CANDIDATE ANALYSIS"
)

print(
    "=" * 80
)


for i in range(
    min(
        10,
        len(df)
    )
):

    candidate = df.iloc[i]


    print(
        "\n" + "-" * 80
    )


    print(
        "Rank:",
        candidate["Rank"]
    )


    print(
        "Candidate ID:",
        candidate["ID"]
    )


    print(
        "Category:",
        candidate["Category"]
    )


    print(
        "TF-IDF Similarity:",
        f"{candidate['similarity_score']:.2f}%"
    )


    print(
        "Skill Match:",
        f"{candidate['skill_match_score']:.2f}%"
    )


    print(
        "Final Score:",
        f"{candidate['final_score']:.2f}%"
    )


    print(
        "Recommendation:",
        candidate["recommendation"]
    )


    print(
        "\nMatched Skills:"
    )


    if candidate["matched_skills"]:

        print(
            ", ".join(
                candidate[
                    "matched_skills"
                ]
            )
        )

    else:

        print(
            "None"
        )


    print(
        "\nMissing Skills:"
    )


    if candidate["missing_skills"]:

        print(
            ", ".join(
                candidate[
                    "missing_skills"
                ]
            )
        )

    else:

        print(
            "None"
        )


# ================================================================
# 34. TOP 10 CANDIDATE GRAPH
# ================================================================

top10 = df.head(
    10
).copy()


plt.figure(
    figsize=(13, 8)
)


sns.barplot(
    data=top10,
    x="final_score",
    y=top10["ID"].astype(str)
)


plt.title(
    "Top 10 Candidates Based on Final Score",
    fontsize=16
)


plt.xlabel(
    "Final Score (%)"
)


plt.ylabel(
    "Candidate ID"
)


plt.xlim(
    0,
    100
)


plt.tight_layout()


plt.show()


# ================================================================
# 35. SIMILARITY VS SKILL MATCH GRAPH
# ================================================================

x = np.arange(
    len(top10)
)


width = 0.35


plt.figure(
    figsize=(15, 8)
)


plt.bar(
    x - width / 2,
    top10["similarity_score"],
    width,
    label="TF-IDF Similarity"
)


plt.bar(
    x + width / 2,
    top10["skill_match_score"],
    width,
    label="Skill Match"
)


plt.xlabel(
    "Candidate ID"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "TF-IDF Similarity vs Skill Match"
)


plt.xticks(
    x,
    top10["ID"].astype(str),
    rotation=45
)


plt.ylim(
    0,
    100
)


plt.legend()


plt.tight_layout()


plt.show()


# ================================================================
# 36. RECOMMENDATION DISTRIBUTION
# ================================================================

recommendation_counts = (
    df[
        "recommendation"
    ]
    .value_counts()
)


plt.figure(
    figsize=(10, 6)
)


sns.barplot(
    x=recommendation_counts.index,
    y=recommendation_counts.values
)


plt.title(
    "Candidate Recommendation Distribution"
)


plt.xlabel(
    "Recommendation"
)


plt.ylabel(
    "Number of Candidates"
)


plt.tight_layout()


plt.show()


# ================================================================
# 37. CATEGORY PERFORMANCE
# ================================================================

category_scores = (
    df.groupby(
        "Category"
    )[
        "final_score"
    ]
    .mean()
    .sort_values(
        ascending=False
    )
    .head(
        15
    )
)


plt.figure(
    figsize=(14, 8)
)


sns.barplot(
    x=category_scores.values,
    y=category_scores.index
)


plt.title(
    "Average Candidate Score by Resume Category"
)


plt.xlabel(
    "Average Final Score (%)"
)


plt.ylabel(
    "Resume Category"
)


plt.xlim(
    0,
    100
)


plt.tight_layout()


plt.show()


# ================================================================
# 38. MISSING SKILLS ANALYSIS
# ================================================================

missing_skill_counter = Counter()


for skills in df[
    "missing_skills"
]:

    missing_skill_counter.update(
        skills
    )


missing_skill_df = pd.DataFrame(
    missing_skill_counter.most_common(
        15
    ),
    columns=[
        "Skill",
        "Number_of_Candidates"
    ]
)


print(
    "\n" + "=" * 80
)

print(
    "MOST COMMON MISSING SKILLS"
)

print(
    "=" * 80
)


if not missing_skill_df.empty:

    display(
        missing_skill_df
    )

else:

    print(
        "No missing skills detected."
    )


# ================================================================
# 39. MISSING SKILLS GRAPH
# ================================================================

if not missing_skill_df.empty:

    plt.figure(
        figsize=(14, 8)
    )


    sns.barplot(
        data=missing_skill_df,
        x="Number_of_Candidates",
        y="Skill"
    )


    plt.title(
        "Most Common Missing Skills"
    )


    plt.xlabel(
        "Number of Candidates"
    )


    plt.ylabel(
        "Missing Skill"
    )


    plt.tight_layout()


    plt.show()


# ================================================================
# 40. RECRUITER-FRIENDLY FINAL TABLE
# ================================================================

final_results = df[
    [
        "Rank",
        "ID",
        "Category",
        "similarity_score",
        "skill_match_score",
        "final_score",
        "matched_skills",
        "missing_skills",
        "recommendation"
    ]
].copy()


# Convert skill lists to strings
final_results[
    "matched_skills"
] = final_results[
    "matched_skills"
].apply(
    lambda x:
    ", ".join(x)
    if x
    else "None"
)


final_results[
    "missing_skills"
] = final_results[
    "missing_skills"
].apply(
    lambda x:
    ", ".join(x)
    if x
    else "None"
)


# Round scores
for column in [
    "similarity_score",
    "skill_match_score",
    "final_score"
]:

    final_results[
        column
    ] = final_results[
        column
    ].round(
        2
    )


print(
    "\n" + "=" * 80
)

print(
    "FINAL RECRUITER-FRIENDLY RESULTS"
)

print(
    "=" * 80
)


display(
    final_results.head(
        20
    )
)


# ================================================================
# 41. SAVE CSV
# ================================================================

CSV_OUTPUT = (
    "ranked_candidates.csv"
)


final_results.to_csv(
    CSV_OUTPUT,
    index=False
)


print(
    "\n✓ CSV output created:"
)


print(
    os.path.abspath(
        CSV_OUTPUT
    )
)


# ================================================================
# 42. SAVE EXCEL
# ================================================================

EXCEL_OUTPUT = (
    "ranked_candidates.xlsx"
)


try:

    final_results.to_excel(
        EXCEL_OUTPUT,
        index=False
    )


    print(
        "\n✓ Excel output created:"
    )


    print(
        os.path.abspath(
            EXCEL_OUTPUT
        )
    )


except Exception as e:

    print(
        "\nExcel export could not be completed."
    )

    print(
        "Reason:",
        e
    )


# ================================================================
# 43. BEST CANDIDATE REPORT
# ================================================================

best_candidate = df.iloc[
    0
]


print(
    "\n" + "=" * 80
)

print(
    "BEST MATCHED CANDIDATE"
)

print(
    "=" * 80
)


print(
    "\nRank:",
    best_candidate[
        "Rank"
    ]
)


print(
    "Candidate ID:",
    best_candidate[
        "ID"
    ]
)


print(
    "Resume Category:",
    best_candidate[
        "Category"
    ]
)


print(
    "TF-IDF Similarity:",
    f"{best_candidate['similarity_score']:.2f}%"
)


print(
    "Skill Match:",
    f"{best_candidate['skill_match_score']:.2f}%"
)


print(
    "Final Score:",
    f"{best_candidate['final_score']:.2f}%"
)


print(
    "Recommendation:",
    best_candidate[
        "recommendation"
    ]
)


print(
    "\nMatched Skills:"
)


if best_candidate[
    "matched_skills"
]:

    print(
        ", ".join(
            best_candidate[
                "matched_skills"
            ]
        )
    )

else:

    print(
        "None"
    )


print(
    "\nMissing Skills:"
)


if best_candidate[
    "missing_skills"
]:

    print(
        ", ".join(
            best_candidate[
                "missing_skills"
            ]
        )
    )

else:

    print(
        "None"
    )


# ================================================================
# 44. FINAL PROJECT SUMMARY
# ================================================================

print(
    "\n\n" + "=" * 80
)

print(
    "FUTURE INTERNS - MACHINE LEARNING TASK 3"
)

print(
    "RESUME / CANDIDATE SCREENING SYSTEM"
)

print(
    "=" * 80
)


print(
    f"""
PROJECT STATUS
--------------

✓ Dataset loaded successfully
✓ Resume text cleaned
✓ NLP preprocessing completed
✓ Skills extracted from resumes
✓ Job description analyzed
✓ Required job skills extracted
✓ TF-IDF vectorization completed
✓ Cosine similarity calculated
✓ Skill matching completed
✓ Missing skills identified
✓ Candidate scores calculated
✓ Candidates ranked
✓ Candidate recommendations generated
✓ Visualizations generated
✓ CSV report generated
✓ Excel report generated


DATASET
-------

Dataset File       : Resume.csv
Total Resumes      : {len(df)}
Total Categories   : {df["Category"].nunique()}


JOB ROLE
--------

Data Scientist


SCORING MODEL
-------------

TF-IDF Similarity : 60%
Skill Match       : 40%


FORMULA
-------

Final Score =
(0.60 × TF-IDF Similarity)
+
(0.40 × Skill Match)


RECOMMENDATIONS
---------------

75% and above  → Strong Match
55% - 74.99%   → Good Match
35% - 54.99%   → Moderate Match
Below 35%      → Low Match


OUTPUT FILES
------------

ranked_candidates.csv
ranked_candidates.xlsx


IMPORTANT
---------

This system is a decision-support tool for recruiters.
The ranking should assist human review and should not be
used as the sole basis for an employment decision.
"""
)


print(
    "=" * 80
)

print(
    "PROJECT COMPLETED SUCCESSFULLY!"
)

print(
    "=" * 80
)
```
